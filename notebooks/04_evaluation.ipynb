```python
import numpy as np
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from rouge_score import rouge_scorer
```

```python
# Indian domain test pairs [question, reference answer]
TEST_PAIRS = [
    ('What is IPC Section 302?', 'IPC Section 302 deals with punishment for murder — death penalty or life imprisonment plus fine.'),
    ('Best budget hotel in Kolkata under 2000?', 'Broadway Hotel and Hotel Centrum are good budget options in Kolkata under 2000 rupees.'),
    ('What does Bhagavad Gita say about karma?', 'The Gita teaches performing duty without attachment to fruits of action — this is Karmayoga.'),
    ('Victoria Memorial entry fee?', 'Entry fee for Victoria Memorial is 30 rupees for Indians and 500 rupees for foreigners.'),
    ('What is Article 21?', 'Article 21 guarantees the Right to Life and Personal Liberty to all persons in India.'),
    ('How to reach Dakshineswar from Howrah?', 'Take Circular Railway from Sealdah station to Dakshineswar, approximately 40 minutes and 5 to 10 rupees.'),
]

# Simulated Sarathi predictions (replace with real model outputs after training)
PREDICTIONS = [
    'Section 302 of the Indian Penal Code prescribes punishment for murder, including life imprisonment or death penalty and fine.',
    'For budget hotels in Kolkata under 2000 rupees, Broadway Hotel on Ganesh Chandra Avenue and Hotel Centrum on Marquis Street are popular choices.',
    'The Bhagavad Gita in Chapter 3 teaches the principle of Karmayoga — performing one duty without being attached to the results.',
    'Victoria Memorial charges 30 rupees entry for Indian nationals and 500 rupees for foreign tourists.',
    'Article 21 of the Indian Constitution protects the right to life and personal liberty of every person.',
    'From Howrah, you can reach Dakshineswar Temple by taking the Circular Railway from Sealdah, which takes about 40 minutes and costs around 5 rupees.',
]
```

```python
smoother = SmoothingFunction().method1
scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

bleu_all, r1_all, r2_all, rl_all = [], [], [], []

print('=' * 70)
for i, (pred, (q, ref)) in enumerate(zip(PREDICTIONS, TEST_PAIRS), 1):
    bleu = sentence_bleu([ref.split()], pred.split(), smoothing_function=smoother)
    rouge = scorer.score(ref, pred)
    bleu_all.append(bleu)
    r1_all.append(rouge['rouge1'].fmeasure)
    r2_all.append(rouge['rouge2'].fmeasure)
    rl_all.append(rouge['rougeL'].fmeasure)
    print(f'Test {i}: {q}')
    print(f'  BLEU={bleu:.4f} | ROUGE-1={rouge["rouge1"].fmeasure:.4f} | ROUGE-2={rouge["rouge2"].fmeasure:.4f} | ROUGE-L={rouge["rougeL"].fmeasure:.4f}')
    print()

print('=' * 70)
print(f'Average BLEU   : {np.mean(bleu_all):.4f}  (target > 0.30)')
print(f'Average ROUGE-1: {np.mean(r1_all):.4f}  (target > 0.50)')
print(f'Average ROUGE-2: {np.mean(r2_all):.4f}  (target > 0.25)')
print(f'Average ROUGE-L: {np.mean(rl_all):.4f}  (target > 0.40)')
```

3. Rerun the evaluation cell.
